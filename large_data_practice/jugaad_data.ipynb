1. Calculate the daily return percentage of TCS stock for each trading day.
- Formula:
-    Daily Return(%)= PREV.CLOSE
-                    ------------- ×100
                     CLOSE−PREV.CLOSE
	​

```python
df["Daily_Return_%"] =(
    (df["CLOSE"] - df["PREV. CLOSE"])
    / df["PREV. CLOSE"]
) * 100
```

```python
df
```

```python
df["Daily_Return_%"] = df["Daily_Return_%"].round(2)
```

Calculate the average closing price of TCS for each month.

```python
df["Month"] = df["DATE"].dt.to_period("M")

Monthly_Avg_Close = df.groupby("Month")["CLOSE"].mean()

Monthly_Avg_Close
```

```python
plt.figure(figsize=(12, 6), facecolor="lightgray")

bars = plt.bar(
    Monthly_Avg_Close.index.astype(str),
    Monthly_Avg_Close.values,
    color="steelblue",
    edgecolor="black",
    linewidth=1.2
)

plt.title(
    "TCS Monthly Average Closing Price",
    fontsize=20,
    fontweight="bold",
    pad=20
)

plt.xlabel(
    "Month",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel(
    "Average Closing Price (₹)",
    fontsize=13,
    fontweight="bold"
)

plt.xticks(
    rotation=45,
    ha="right",
    fontsize=10
)

plt.yticks(fontsize=10)

plt.gca().set_facecolor("white")

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

# Value labels
for bar in bars:
    value = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"₹{value:,.0f}",
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold"
    )

plt.ylim(
    0,
    Monthly_Avg_Close.max() * 1.15
)

plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```

Q14. Maximum Drawdown
- Calculate the Maximum Drawdown of TCS closing price to identify the maximum percentage fall from a previous peak.

```python
id="q14calc"
df["Peak"] = df["CLOSE"].cummax()

df["Drawdown_%"] = (
    (df["CLOSE"] - df["Peak"])
    / df["Peak"]
) * 100

Max_Drawdown = df["Drawdown_%"].min()

Max_Drawdown
```

```python
plt.figure(figsize=(12, 6), facecolor="lightgray")

plt.plot(
    df["DATE"],
    df["Drawdown_%"],
    linewidth=2
)

plt.axhline(
    0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "TCS Drawdown from Previous Peak",
    fontsize=18,
    fontweight="bold",
    pad=15
)

plt.xlabel(
    "Date",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel(
    "Drawdown (%)",
    fontsize=13,
    fontweight="bold"
)

plt.xticks(rotation=45)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

plt.gca().set_facecolor("white")
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```

Q15. Final TCS Performance Summary
- Prepare a final summary of TCS stock performance by calculating the overall price change, average daily return, maximum daily gain, maximum daily loss, and maximum drawdown.

```python
Overall_Price_Change = (
    (df["CLOSE"].iloc[-1] - df["CLOSE"].iloc[0])
    / df["CLOSE"].iloc[0]
) * 100

Average_Daily_Return = df["Daily_Return_%"].mean()

Maximum_Daily_Gain = df["Daily_Return_%"].max()

Maximum_Daily_Loss = df["Daily_Return_%"].min()

Maximum_Drawdown = df["Drawdown_%"].min()

Summary = pd.Series({
    "Overall Price Change (%)": round(Overall_Price_Change, 2),
    "Average Daily Return (%)": round(Average_Daily_Return, 2),
    "Maximum Daily Gain (%)": round(Maximum_Daily_Gain, 2),
    "Maximum Daily Loss (%)": round(Maximum_Daily_Loss, 2),
    "Maximum Drawdown (%)": round(Maximum_Drawdown, 2)
})

Summary
```
