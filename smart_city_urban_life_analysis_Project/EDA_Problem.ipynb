### NumPy Analysis

Q1. Create a NumPy array for Temperature_C and find the minimum, maximum, and mean temperature.

```python
temp = dc["Temperature_C"].to_numpy()
# # Minimum temperature
min_temp = np.min(temp)

# # Maximum temperature
max_temp = np.max(temp)

# # Mean temperature
mean_temp = np.mean(temp).round(1)

print("Minimum Temperature:", min_temp)
print("Maximum Temperature:", max_temp)
print("Mean Temperature:", mean_temp)
```

Q2. Calculate the mean, median, and standard deviation of AQI(Air Quality Index).

```python
dc["AQI"].head(5)
```

```python
AQI = dc["AQI"].to_numpy()

# Mean Air Quality Index
MEAN_AQI = np.mean(AQI).round(1)

# Median Air Quality Index
MEDIAN_AQI = np.median(AQI)

# Standard deviation Air Quality Index
STD_AQI = np.std(AQI).round(1)

print("Mean Air Quality Index:", MEAN_AQI)
print("Median Air Quality Index:", MEDIAN_AQI)
print("Standard deviation Air Quality Index:", STD_AQI)
```

Q8. Calculate the average temperature for each city.

```python
avg_temp = dc.groupby("City")["Temperature_C"].mean().round(2)
avg_temp
```

### Q2. City-wise Average Temperature
Create a Bar Chart to compare the average temperature across all cities.

```python
avg_temp = avg_temp.sort_values(ascending=False)
plt.figure(figsize=(13, 7), facecolor="#121314")
bars = plt.bar(
    avg_temp.index,
    avg_temp.values,
    color=["green", "white", "blue", "brown", "Yellow", "red", "cyan", "pink"],
    edgecolor="black",
    linewidth=1.2
)

plt.title(
    "CITY WISE AVERAGE TEMPERATURE",
    color="white",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.xlabel(
    "CITY",
    color="white",
    fontsize=13,
    fontweight="bold",
    labelpad=13
)

plt.ylabel(
    "TEMPERATURE",
    color="white",
    fontsize=13,
    fontweight="bold",
    labelpad=13 
)

plt.xticks(
    color="white",
    ha="center",
    rotation=0,
    fontweight="bold", 
)

plt.yticks(
    color="white",
    fontweight="bold", 
    fontsize=13
)

plt.grid(
    axis="y",
    linestyle="--",
    color="white",
    alpha=0.9,
)

for bar in bars:
    value = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 1,
        f"{value:,.2f}",
        ha="center",
        va="bottom",
        fontsize=13,
        fontweight="bold",
        color="white"
    )

plt.gca().set_facecolor("#121314")
plt.gca().spines["left"].set_color("cyan")
plt.gca().spines["bottom"].set_color("blue")
plt.gca().spines["top"].set_color("black")
plt.gca().spines["right"].set_color("black")

plt.tight_layout()

plt.show()
```
