### NumPy Analysis

Q1. Create a NumPy array for Temperature_C and find the minimum, maximum, and mean temperature.

```python
temp = dc["Temperature_C"].to_numpy()
# # Minimum temperature
min_temp = np.min(temp)

# # Maximum temperature
max_temp = np.max(temp)

# # Mean temperature
mean_temp = np.mean(temp).round(1)

print("Minimum Temperature:", min_temp)
print("Maximum Temperature:", max_temp)
print("Mean Temperature:", mean_temp)
```

Q2. Calculate the mean, median, and standard deviation of AQI(Air Quality Index).

```python
dc["AQI"].head(5)
```

```python
AQI = dc["AQI"].to_numpy()

# Mean Air Quality Index
MEAN_AQI = np.mean(AQI).round(1)

# Median Air Quality Index
MEDIAN_AQI = np.median(AQI)

# Standard deviation Air Quality Index
STD_AQI = np.std(AQI).round(1)

print("Mean Air Quality Index:", MEAN_AQI)
print("Median Air Quality Index:", MEDIAN_AQI)
print("Standard deviation Air Quality Index:", STD_AQI)
```

## 🕊️ Bar Charts

```python
Avg_Aqi_City = Avg_Aqi_City.sort_values(ascending=False)
plt.figure(figsize=(12, 6), facecolor="#121314")

bar = plt.bar(
    Avg_Aqi_City.index,
    Avg_Aqi_City.values,
    color=["green", "Yellow", "red", "cyan", "pink", "white", "blue", "brown"],
    edgecolor="black",
    linewidth=1.2
)

plt.title(
    "CITY WISE AIR QUALITY INDEX",
    color="white",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.xlabel(
    "City",
    color="white",
    fontsize = 18,
    fontweight="bold",
    labelpad=13
)

plt.ylabel(
    "Average AQI",
    color="white",
    fontsize = 18,
    fontweight="bold",
    labelpad=13
)

plt.gca().set_facecolor("#121314")

plt.xticks(
    rotation=0,
    ha="center",
    fontsize=10,
    fontweight="bold",
    color="white"
)

plt.yticks(
    fontsize=10,
    fontweight="bold",
    color="white"
)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.4,
    color="cyan"
)

for bar in bar:
    value = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:,.2f}",
        ha="center",
        va="bottom",
        fontsize=13,
        fontweight="bold",
        color="white"
    )

plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)
plt.gca().spines["left"].set_color("red")
plt.gca().spines["bottom"].set_color("yellow")

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.4,
    color="cyan"
)

plt.gca().xaxis.grid(False)

plt.tight_layout()

plt.show()
```

## Weather Analysis

Q13. Calculate the average temperature for each month Name.

```python
avg_month_temp = (
    dc.groupby(["Month", "Month_Name"])["Temperature_C"]
    .mean()
    .round(2)
    .reset_index()
    .sort_values("Month")
    .reset_index(drop=True)
)

avg_month_temp
```

Q14. Calculate the total rainfall for each month.

```python
Total_rainfall_month =(
    dc.groupby(["Month", "Month_Name"])["Rainfall_mm"]
    .sum()
    .reset_index()
    .sort_values("Month")
)

Total_rainfall_month
```

Q15. Which month has the highest average temperature?

```python
avg_month_temp = (
    dc.groupby("Month_Name")["Temperature_C"]
    .mean()
    .round(2)
)

highest_month = avg_month_temp.idxmax()

print("Month Name with Highest Average Temperature:", highest_month)
print("Highest Average Temperature:", avg_month_temp.max())
```

Q16. Which month has the highest total rainfall?

```python
highest_month_rainfall = (
    dc.groupby("Month_Name")["Rainfall_mm"]
    .sum()
    .round(2)
)

highest_total_rainfall = highest_month_rainfall.idxmax()

print("Month Name with Highest Total Rainfall:", highest_total_rainfall)
print("Highest Rainfall:", highest_month_rainfall.max())
```
