### NumPy Analysis

Q1. Create a NumPy array for Temperature_C and find the minimum, maximum, and mean temperature.

```python
temp = dc["Temperature_C"].to_numpy()
# # Minimum temperature
min_temp = np.min(temp)

# # Maximum temperature
max_temp = np.max(temp)

# # Mean temperature
mean_temp = np.mean(temp).round(1)

print("Minimum Temperature:", min_temp)
print("Maximum Temperature:", max_temp)
print("Mean Temperature:", mean_temp)
```

Q2. Calculate the mean, median, and standard deviation of AQI(Air Quality Index).

```python
dc["AQI"].head(5)
```

```python
AQI = dc["AQI"].to_numpy()

# Mean Air Quality Index
MEAN_AQI = np.mean(AQI).round(1)

# Median Air Quality Index
MEDIAN_AQI = np.median(AQI)

# Standard deviation Air Quality Index
STD_AQI = np.std(AQI).round(1)

print("Mean Air Quality Index:", MEAN_AQI)
print("Median Air Quality Index:", MEDIAN_AQI)
print("Standard deviation Air Quality Index:", STD_AQI)
```

## 🏙️ Smart City Urban Life Analysis

## 1. City-wise Average Happiness Score
- What is the average Happiness Score for each city?

```python
avg_score = (
    dc.groupby("City")["Happiness_Score"].mean().round(2)
)
avg_score
```

### Bar Chart

```python
plt.figure(figsize=(13, 6), facecolor="#121314")
bars = plt.bar(
    avg_score.index,
    avg_score.values,
    color=["green", "orange", "purple", "red", "pink", "white", "blue", "yellow"],
    
)

plt.gca().set_facecolor("black")

plt.title(
    "City-wise Average Happiness Score",
    color="white",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.xlabel(
    "City",
    color="white",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

plt.ylabel(
    "Happiness Score",
    color="white",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

plt.xticks(
    color="white",
    fontsize=10,
    fontweight="bold"
)

plt.yticks(
    color="white",
    fontsize=10,
    fontweight="bold"
)

plt.gca().spines["left"].set_color("blue")
plt.gca().spines["bottom"].set_color("blue")
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

for bar in bars:
    values = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        values,
        f"{values:.2f}",
        ha="center",
        va="bottom",
        color="white"
    )

plt.show()
```

## 2. City-wise Average AQI
- Which cities have higher average AQI levels?

```python
Avg_city = (
    dc.groupby("City")["AQI"]
      .mean()
      .round(2)
      .sort_values(ascending=False)
      .head(10)
)
Avg_city
```

### Bar Chart

```python
plt.figure(figsize=(13, 6), facecolor="#121314")
bars = plt.bar(
    Avg_city.index,
    Avg_city.values,
    color=["green", "orange", "purple", "red", "pink", "white", "blue", "yellow"],
    
)

plt.gca().set_facecolor("#121314")

plt.title(
    "City-wise Average AQI",
    color="white",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.xlabel(
    "City",
    color="white",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

plt.ylabel(
    "Air Quality Index",
    color="white",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

plt.xticks(
    color="white",
    fontsize=10,
    fontweight="bold"
)

plt.yticks(
    color="white",
    fontsize=10,
    fontweight="bold"
)

plt.gca().spines["left"].set_color("blue")
plt.gca().spines["bottom"].set_color("blue")
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

for bar in bars:
    values = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        values,
        f"{values:.2f}",
        ha="center",
        va="bottom",
        color="white"
    )

plt.show()
```

## 3. Monthly Average AQI Trend
- How does the average AQI change month by month?

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]
avg_month = (
    dc.groupby("Month_Name")["AQI"]
    .mean()
    .round(2)
    .sort_index()
    .head(12)
    .reindex(month_order)
)
avg_month
```

### Line Chart

```python
plt.figure(figsize=(15, 6), facecolor="#121314")

# --------------------------------------------------
# Line Chart
# --------------------------------------------------

plt.plot(
    avg_month.index,
    avg_month.values,
    marker="o",
    markersize=7,
    markeredgecolor="white",
    markeredgewidth=1.5,
    color="#FFD700",
    linewidth=2.5
)


# --------------------------------------------------
# Title
# --------------------------------------------------

plt.title(
    "Monthly Average AQI Trend",
    fontsize=20,
    fontweight="bold",
    color="white",
    pad=20
)


# --------------------------------------------------
# X-axis Label
# --------------------------------------------------

plt.xlabel(
    "Month Name",
    labelpad=13,
    fontsize=14,
    color="white",
    fontweight="bold"
)


# --------------------------------------------------
# Y-axis Label
# --------------------------------------------------

plt.ylabel(
    "Air Quality Index (AQI)",
    labelpad=13,
    fontsize=14,
    color="white",
    fontweight="bold"
)


# --------------------------------------------------
# Grid
# --------------------------------------------------

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.25,
    color="white"
)


# --------------------------------------------------
# X-axis Ticks
# --------------------------------------------------

plt.xticks(
    color="white",
    fontsize=10,
    fontweight="bold",
    # rotation=30
)


# --------------------------------------------------
# Y-axis Ticks
# --------------------------------------------------

plt.yticks(
    ticks=range(108, 124, 2),
    color="white",
    fontsize=10,
    fontweight="bold"
)


# --------------------------------------------------
# Chart Background
# --------------------------------------------------

plt.gca().set_facecolor("#3c3c3c")


# --------------------------------------------------
# Value Labels
# --------------------------------------------------

for x, y in zip(avg_month.index, avg_month.values):

    plt.text(
        x,
        y + 0.45,
        f"{y:.2f}",
        ha="center",
        va="bottom",
        color="white",
        fontsize=10,
        fontweight="bold"
    )


# --------------------------------------------------
# Spines
# --------------------------------------------------

plt.gca().spines["left"].set_color("white")
plt.gca().spines["bottom"].set_color("white")

plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)


# --------------------------------------------------
# Layout
# --------------------------------------------------

plt.tight_layout()

plt.show()
```

### 4. Distribution of Happiness Score

### Histogram

```python
plt.figure(figsize=(15, 6), facecolor="#121314")

counts, bins, patches = plt.hist(
    dc["Happiness_Score"],
    bins=20,
    edgecolor="black",
    alpha=0.7
)

# --------------------------------------------------
# Different color for each bar
# --------------------------------------------------

colors = [
    "red", "blue", "green", "yellow", "orange",
    "purple", "pink", "cyan", "lime", "brown",
    "gray", "gold", "teal", "coral", "navy",
    "violet", "indigo", "silver", "magenta", "olive"
]

for i in range(len(patches)):
    patches[i].set_facecolor(colors[i])


# --------------------------------------------------
# Frequency value on top of each bar
# --------------------------------------------------

for i in range(len(counts)):

    plt.text(
        patches[i].get_x() + patches[i].get_width() / 2,
        counts[i] + 200,
        f"{int(counts[i])}",
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold",
        color="white"
    )


# --------------------------------------------------
# Title
# --------------------------------------------------

plt.title(
    "Distribution of Happiness Score",
    color="white",
    fontsize=18,
    fontweight="bold",
    pad=20
)


# --------------------------------------------------
# X-axis
# --------------------------------------------------

plt.xticks(
    [3, 3.5, 4, 4.5, 5, 5.5, 6, 6.5,
     7, 7.5, 8, 8.5, 9, 9.5, 10],
    color="white",
    fontsize=10,
    fontweight="bold"
)


# --------------------------------------------------
# Y-axis
# --------------------------------------------------

plt.yticks(
    color="white",
    fontsize=10,
    fontweight="bold"
)


# --------------------------------------------------
# X-axis Label
# --------------------------------------------------

plt.xlabel(
    "Happiness Score",
    color="white",
    fontsize=18,
    fontweight="bold",
    labelpad=20
)


# --------------------------------------------------
# Y-axis Label
# --------------------------------------------------

plt.ylabel(
    "Frequency",
    color="white",
    fontsize=18,
    fontweight="bold",
    labelpad=20
)


# --------------------------------------------------
# Chart background
# --------------------------------------------------

plt.gca().set_facecolor("#3c3c3c")


# --------------------------------------------------
# Grid
# --------------------------------------------------

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.25,
    color="white"
)


# --------------------------------------------------
# Remove unnecessary borders
# --------------------------------------------------

plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

plt.gca().spines["left"].set_color("white")
plt.gca().spines["bottom"].set_color("white")


# --------------------------------------------------
# Final layout
# --------------------------------------------------

plt.tight_layout()

plt.show()
```
