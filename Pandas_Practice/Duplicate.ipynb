# 🔥 Duplicate Data in pandas

### 📌 What is Duplicate Data?
Duplicate data means same rows or values appearing more than once in a dataset

### 1️⃣ Detect Duplicate Data
1. duplicated()
- Checks whether a row is duplicate or not.

# 🧠 Syntax

```python
df.duplicated()
```

### ✨ Example

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['A', 'B', 'A', 'C', 'B'],
    'Marks': [50, 60, 50, 70, 60]
})

print(df.duplicated())
```

# 2️⃣ Remove Duplicate Rows
### 📌 What is `drop_duplicates()`
Removes duplicate rows from DataFrame.

# 🧠 Syntax

```python
df.drop_duplicates()
```

# 3️⃣ Remove Duplicates from Specific Columns

```python
df.drop_duplicates(subset=['Name'])
```

# 4️⃣ Keep First / Last Duplicate
- 🔹 Keep first (default)

```python
df.drop_duplicates(keep='first')
```

🔹 Keep last

```python
df.drop_duplicates(keep='last')
```

🔹 Remove all duplicates

```python
df.drop_duplicates(keep=False)
```

5️⃣ Remove Duplicates In-Place

```python
df.drop_duplicates(inplace=True)
```

7️⃣ Detect Duplicate Values in a Column

```python
df['Name'].duplicated()
```

8️⃣ Count Duplicate Rows

```python
df.duplicated().sum()
```
