# 🔹 Sorting Data in pandas

# 📌 What is Sorting Data in Pandas?

Sorting data in Pandas means arranging the rows of a DataFrame or Series in a specific order based on:

- column values
- index values

🔹 Why do we sort data?
- To analyze data easily
- To find highest / lowest values
- To make reports clean and readable

```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Marks": [78, 45, 90, 60]
})
```

# 🔹 Types of Sorting in Pandas

### 1️⃣ Sort by Values (Column-based sorting)
Used when you want to sort rows based on column values.

`sort_values()`

```python
df.sort_values(by="Marks")
```

# 2️⃣ Sort by Index
Used when you want to sort rows based on index.
- sort_index()`

```python
df.sort_index()
```

# 🔹 Sort data in ascending order (default)

```python
df.sort_values(by="Marks")
```

🔹 Sort data in descending order

```python
df.sort_values(by="Marks", ascending=False)
```

🔹 Sort by multiple columns

```python
df.sort_values(by=["Marks", "Name"])
```

🔹 Sorting with missing values

```python
df.sort_values(by="Marks", na_position="last")
```
