# Handle Missing Value

```python
df.dropna(subset=["emp_id"], inplace=True)
```

```python
df['age'] = (
    df['age']
    .apply(lambda x: int(round(x)) if pd.notna(x) else None)
    .astype('Int64')
)
```

```python
df["age"].dtype
df.head()
```

```python
df['salary'] = (
    df['salary']
    .apply(lambda x: int(round(x)) if pd.notna(x) else None)
    .astype('Int64')
)
```

✅Sirf Date (YYYY-MM-DD) rakhna

```python
df["joining_date"] = pd.to_datetime(df["joining_date"]).dt.date
```

String format me clean date

```python
df["joining_date"] = pd.to_datetime(df["joining_date"]).dt.strftime("%d-%m-%Y")
```

```python
df.head(3)
```

Fill missing valuue in rendom number

```python
df['age'] = df['age'].apply(
    lambda x: np.random.randint(22, 35) if pd.isna(x) else x
)
```

```python
df.describe()
```

```python
df['salary'] = df['salary'].apply(
    lambda x: np.random.randint(3000, 6000) if pd.isna(x) else x
)
```

```python
df['name'] = df['name'].apply(
    lambda x: np.random.choice(df['name'].dropna()) if pd.isna(x) else x
)
```

```python
df['department'] = df['department'].apply(
    lambda x: np.random.choice(df['department'].dropna()) if pd.isna(x) else x
)
```

```python
df['city'] = df['city'].apply(
    lambda x: np.random.choice(df['city'].dropna()) if pd.isna(x) else x
)
```
