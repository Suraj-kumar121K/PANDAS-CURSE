# 1️⃣ Melt in pandas

# 1️⃣ What is melt?
melt in pandas is used to convert a DataFrame from wide format to long format.
- Wide format: each variable has its own column
- Long format: each variable becomes a row, with key-value pairs

# 2️⃣ Syntax
Parameter:
- `frame` --> The DataFrame to transform
- `id_vars` --> Columns to keep
- `value_vars` --> Columns to convert into rows
- `var_name` --> Name of the new column for the variable names
- `value_name` --> Name of the new column for the values

```python
import pandas as pd

data = {
    'Department': ['HR', 'IT', 'Sales'],
    'F': [50000, 0, 55000],
    'M': [60000, 150000, 65000]
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

# Using melt
melted = pd.melt(
    df,
    id_vars=['Department'],
    value_vars=['F', 'M'],
    var_name='Gender',
    value_name='Salary'
)
print("\nMelted DataFrame:\n", melted)
```

# 1️⃣ What is a PivotTable?
- A PivotTable in pandas is used to reshape data and aggregate values.

- data by rows (index), columns, and aggregation functions (sum, mean, count, etc.)

- data analysis and reporting.

# Syntax

- pd.pivot_table(
-    data, 
-    values='column_to_aggregate',   # column to summarize
-    index='column_for_rows',        # row labels
-    columns='column_for_columns',   # column labels
-    aggfunc='sum'                   # aggregation function (sum, mean, count, etc.)
- )

# Parameter:
- `data`   --> The DataFrame to pivot
- `values` --> Column to aggregate
- `index`  --> Column to use for row labels
- `columns` --> Column to use for column labels
- `aggfunc` --> Aggregation function

```python
import pandas as pd

data = {
    'Department': ['HR', 'HR', 'IT', 'IT', 'Sales', 'Sales'],
    'Gender': ['F', 'M', 'M', 'M', 'F', 'M'],
    'Salary': [50000, 60000, 70000, 80000, 55000, 65000]
}

df = pd.DataFrame(data)

# Create a PivotTable: sum of Salary by Department and Gender
pivot = pd.pivot_table(df, values='Salary', index='Department', columns='Gender', aggfunc='sum')
print(pivot)
```

# Multiple aggregation functions

```python
pivot = pd.pivot_table(df, values='Salary', index='Department', columns='Gender', aggfunc=['sum', 'mean'])
print(pivot)
```
