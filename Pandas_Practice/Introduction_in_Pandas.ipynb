# 1️⃣ Data Structure in Pandas.
- A data structure in Pandas is a way to organize and store data so that it can be efficiently manipulated, analyzed, and retrieved. Pandas provides two main data structures:

# A. Series (1-dimensional)
Series Pandas ka one-dimensional (1D) data structure hota hai jo data + index ke saath values store karta hai.

- A Series is like a single column of data, with an index attached.
- Can store numbers, strings, or dates.

### 👋 Syntax:

```python
s = pd.Series([10, 20, 30, 40])
```

## B. DataFrame (2-dimensional)
Definition:
- A DataFrame is a 2-dimensional labeled data structure in Pandas. Think of it like a table in Excel or a SQL table: it has rows and columns, where each column can have a different data type.

- A DataFrame is like a table (Excel sheet) with rows and columns.
- Each column is a Series.

# C. Panel(3-dimensional)
Panel pandas ka ek 3-dimensional (3D) data structure tha.
# Store
- items → multiple DataFrames (like variables / sheets)
- major_axis → rows (index)
- minor_axis → columns

# Structure

Panel
`├── Item 1 → DataFrame `
`├── Item 2 → DataFrame`
`└── Item 3 → DataFrame`

## Creating a DataFrame
- A. From a Dictionary

# Syntax:

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['Delhi', 'Mumbai', 'Bangalore']
}

df = pd.DataFrame(data)
print(df)
```

# From a List of Lists

```python
data = [
    ['Alice', 25, 'Delhi'],
    ['Bob', 30, 'Mumbai'],
    ['Charlie', 35, 'Bangalore']
]

df = pd.DataFrame(data, columns=['Name', 'Age', 'City'])
print(df)
```

# 🔹 Series with Custom Index

```python
s = pd.Series([100, 200, 300], index=['a', 'b', 'c'])
print(s)
```

# Series

```python
data = pd.Series([10, None, 10], index=["a", "b", "c"], dtype="float")
print(data)
print(type(data))
```
