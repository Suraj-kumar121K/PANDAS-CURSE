# 🔹 What is nrows in Pandas?

- nrows is a parameter used in `pd.read_csv()` (and read_excel) to read only a fixed number of rows from a file.

### 🔹 Syntax
`pd.read_csv("file.csv", nrows=number)`

# 🔹 What is `usecols`?
- usecols is used with `pd.read_csv()` / pd.`read_excel()` to read only selected columns from a file.

### 🔹 Syntax
-`pd.read_csv("file.csv", usecols=[...])`

pd.read_csv("file.csv", nrows=number)

```python
var = pd.read_excel(r"C:\EXCEL\converted_excel.xlsx", usecols=[0])
var
```

# 🔹 What is header in Pandas?

- header tells Pandas which row should be used as column names.

### header=0
`df = pd.read_csv("data.csv", header=0)`

```python
var = pd.read_excel(r"C:\EXCEL\converted_excel.xlsx", header=2)
var
```

# 🔹 What is names in Pandas?
-names is used to assign column names manually while reading a file.

# Syntax
`pd.read_csv("file.csv", names=[...])`

```python
var = pd.read_excel(r"C:\EXCEL\converted_excel.xlsx", names=["co1", "col2", "col3", "col4"])
var
```

```python
var.to_csv("converted_excel.xlsx", index=False)
```
