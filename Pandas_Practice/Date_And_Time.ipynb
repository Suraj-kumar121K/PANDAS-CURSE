# 📅 Date & Time Handling (Time Series) in pandas

### 📌 What is Time Series Data?
Data where each observation is associated with a date or time.

# 📈 Examples:
- Stock prices
- Sales per day
- Sensor data
- Website traffic

# 1️⃣ Date & Time Data Types in pandas

 `datetime64[ns]` --> Date + Time                      
 `timedelta64[ns]` --> Difference between dates         
 `Period` --> Time span (month, year, quarter) 


# 2️⃣ Convert to Datetime – `to_datetime()`

```python
import pandas as pd

df = pd.DataFrame({
    'Date': ['2024-01-01', '2024-01-05', '2024-02-10']
})

df['Date'] = pd.to_datetime(df['Date'])
print(df.dtypes)
```

# 3️⃣ Extract Date Components – .dt accessor

- `df['Year'] = df['Date'].dt.year`
- `df['Month'] = df['Date'].dt.month`
- `df['Day'] = df['Date'].dt.day`
- `df['Weekday'] = df['Date'].dt.day_name()`

# 4️⃣ Set Date as Index
`df.set_index('Date', inplace=True)`

# 5️⃣ Date Range – `date_range()`

```python
dates = pd.date_range(start='2024-01-01', periods=5, freq='D')
print(dates)
```

# 🔹 Common Frequencies
 `D` -->  Daily       
 `M` -->  Month end   
 `MS`-->  Month start 
 `Y` -->  Year end    
 `H` -->  Hourly      


# 6️⃣ Time-Based Filtering 🔥
- `df.loc['2024-01-01':'2024-01-10']`
- `df.loc['2024-01-01':'2024-01-10']`
