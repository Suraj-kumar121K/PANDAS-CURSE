# ➕ Arithmetic Operators in Pandas
- Pandas supports element-wise arithmetic operations on Series and DataFrames.

# 🔢 List of Arithmetic Operators

- add() --> `+` --> Addition         
- sub() -->  `-` --> Subtraction      
- mul() --> `*` --> Multiplication   
- div() -->  `/` --> Division         
- floordiv() -->  `//` --> Floor Division   
- mod() -->  `%` --> Modulus          
- pow() -->  `**` --> Power (Exponent) 


```python
import pandas as pd

s1 = pd.Series([10, 20, 30])
s2 = pd.Series([2, 4, 6])
```

# ➕ Addition

```python
s1 + s2
```

# ➖ Subtraction

```python
s1 - s2
```

# ✖ Multiplication

```python
s1 * s2
```

# ➗ Division

```python
s1 / s2
```

# ➗➗ Floor Division

```python
s1 // s2
```

# 🔁 Modulus

```python
s1 % s2
```

# ⬆ Power

```python
s1 ** s2
```

# 🗑️ Delete and Insert Data in Pandas
- we can insert (add) or delete (remove) data in rows and columns of a DataFrame.

1️⃣ Insert a New Column
- Insert in Pandas means adding new data (a row or a column) into a DataFrame.

```python
var = pd.DataFrame({"A":[1, 2, 3, 4, 5], "B":[10, 9, 8, 7, 6, ]})
print(var)
```

```python
var.insert(1,"python",var["A"])
var
```
