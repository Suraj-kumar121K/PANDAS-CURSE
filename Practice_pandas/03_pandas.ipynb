```python
data = {
    "Emp_ID": range(1, 51),
    "Name": [f"Employee_{i}" for i in range(1, 51)],
    "Age": np.random.randint(20, 50, 50),
    "Department": np.random.choice(["IT", "HR", "Sales", "Finance"], 50),
    "Salary": np.random.randint(20000, 100000, 50),
    "Experience_Years": np.random.randint(1, 20, 50)
}
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df
```

```python
df.shape
```

```python
df.duplicated().sum()
```

```python
df["Bonus"] = df["Salary"] * 10
df
```

```python
df["Total_Income"] = df["Salary"] + df["Bonus"]
df.head(5)
```

```python
df["Salary_Category"] = np.where(
    df["Salary"] > 60000,
    "High",
    "Low"
)
df.head(5)
```

🔹 4️⃣ Experience Level (apply + lambda)

```python
df["Level"] = df["Experience_Years"].apply(
    lambda x: "Senior" if x > 10 else "Junior"
)
df.head()
```

```python
df["Age_Group"] = df["Age"].apply(
    lambda x: "Young" if x < 30 else "Adult"
)
df
```

```python
df["Dept_Rank"] = df.groupby("Department")["Salary"].rank(ascending=False)
df
```

🔹 8️⃣ Conditional Update (loc Apply)

```python
df.loc[df["Department"] == "IT", "Salary"] += 5000
df
```
