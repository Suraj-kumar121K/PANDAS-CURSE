Handle missing value
- Age

```python
df["Age"] = df["Age"].replace("Unknown", np.nan)
```

```python
df["Age"] = pd.to_numeric(df["Age"])
```

```python
df["Age"] = df["Age"].apply(
    lambda x: np.random.randint(20, 50) if pd.isna(x) else x
)
df
```

```python
df["Age"] = df["Age"].astype(int)
df
```

```python
df["Department"] = df["Department"].replace({
    "it": "IT",
    "Hr": "HR"
})
```

Salary

```python
df["Salary"] = df["Salary"].apply(
    lambda x : np.random.randint(22541, 83768) if pd.isna(x) else x
)
df
```

```python
df["Salary"] = df['Salary'].astype(int)
df
```

✅ Step 2: "Missing value" ko NaN me convert karo

```python
df["Join_Date"] = df["Join_Date"].replace("Missing value", np.nan)
```

✅ Step 3: Convert to Proper Datetime 

```python
df["Join_Date"] = pd.to_datetime(
    df["Join_Date"],
    errors="coerce",
    dayfirst=True
)

df["Join_Date"] = df["Join_Date"].dt.date
df
```
