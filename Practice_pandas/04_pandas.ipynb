## New Columns

11. Calculate Gross Sales

```python
df["Gross_Sales"] = df["Qty"] * df["Original_Price"]
```

1. Create a new column called Profit.
- Formula: Sales - Cost
- Where Cost = Qty × Original_Price × 70%.

```python
df["Profit"] = df["Sales"] - (df["Qty"] * df["Original_Price"] * 0.70)
```

2. Create a new column called Profit_Margin.
- Formula: (Profit / Sales) × 100.

```python
df["Profit_Margin"] = ((df["Profit"] / df["Sales"]) * 100).apply(int)
```

3. Create a new column called Discount_Amount.
- Formula: Gross_Sales - Sales.

```python
df["Discount_Amount"] = df["Gross_Sales"] - df["Sales"]
```

4. Create a new column called Final_Revenue.
- Formula: Sales - Discount_Amount.

```python
df["Final_Revenue"] = df["Sales"] - df["Discount_Amount"]
```

5. Create a new column called Price_With_GST.
- Add 18% GST to Unit_Price.

```python
df["Price_With_GST"] = (df["Unit_Price"] * 1.18).apply(int)
```

6. Create a new column called Net_Value.
- Formula: Qty × Unit_Price.

```python
df["Net_Value"] = df["Qty"] * df["Unit_Price"]
```

7. Create a Revenue_Category column:
- Sales >= 50,000 → Very High
- Sales >= 25,000 → High
- Sales >= 10,000 → Medium
- Sales >= 5,000 → Low
- Otherwise → Very Low

```python
df["Revenue_Category"] = df["Sales"].apply(
    lambda x:"Very High" if x >= 50000
    else "High" if x >= 25000 
    else "Medium" if x >= 10000
    else "Low" if x >= 5000
    else "Very Low" 
)
```

8. Create a Performance column using both Sales and Discount:
- Sales > 50,000 AND Discount < 10% → Excellent
- Sales > 20,000 AND Discount < 15% → Good
- Sales > 5,000 → Average
- Otherwise → Poor

```python
df["Performance"] = df.apply(
    lambda row: "Excellent" if row["Sales"] > 50000 and row["Discount"] < 0.10
    else "Good" if row["Sales"] > 20000 and row["Discount"] < 0.15
    else "Average" if row["Sales"] > 5000
    else "Poor",
    axis=1
)
```
