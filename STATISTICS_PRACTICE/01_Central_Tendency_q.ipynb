### Data Type Change

```python
date = [
    "Date of Birth",
    "Sales Date",
    "Order Date",
    "Ship Date"
]

df[date] = df[date].apply(pd.to_datetime)
```

### Two column add

```python
df["Full Name"] = (
    df["Customer Name"].fillna("") + " " +
    df["Last Name"].fillna("")
).str.strip()
```

```python
df["Full Name"]
```

### Drop column

```python
df = df.drop(columns=["Customer Name", "Last Name"])
```

### column change

```python
df = df[
    ["Customer ID", "Full Name"] +
    [col for col in df.columns if col not in ["Customer ID", "Full Name"]]
]
```
