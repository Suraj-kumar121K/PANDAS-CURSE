# Merging, Joining & Concatenation in pandas

# 1️⃣ Concatenation 
- Concatenation is basically stacking DataFrames either vertically or horizontally.

Syntax:
- `pd.concat([df1, df2], axis=0, ignore_index=False)`

Parameters:
- `axis=0` → vertical (default)
- `axis=1` → horizontal
- `ignore_index` = True → reset index


```python
import pandas as pd

df1 = pd.DataFrame({'A':[1,2], 'B':[3,4]})
df2 = pd.DataFrame({'A':[5,6], 'B':[7,8]})

# Vertical concatenation
result = pd.concat([df1, df2], axis=0)
print(result)
```

Horizontal concatenation:

```python
result = pd.concat([df1, df2], axis=1)
print(result)
```

# 2️⃣ Merge
Merge is a method to combine two DataFrames into one based on common columns or indices.

### Syntax:
- `pd.merge(df1, df2, on='key', how='inner')`

Parameters:
- `on` → column name to join on
- `how` → type of join: `'inner'`, `'outer'`, `'left'`, `'right'`

```python
df1 = pd.DataFrame({'ID':[1,2,3], 'Name':['Alice','Bob','Charlie']})
df2 = pd.DataFrame({'ID':[2,3,4], 'Age':[25,30,35]})

# Merge on ID (inner join)
result = pd.merge(df1, df2, on='ID', how='inner')
print(result)
```

# 3️⃣ Join
- Join is a way to combine two DataFrames based on their indices by default.

- It’s similar to SQL joins, but instead of columns, it mainly uses the index of the DataFrames.

Syntax:
- `df1.join(df2, how='left', lsuffix='', rsuffix='')`
### Parameters:
- `df2` → DataFrame you want to join with df1

- `how` → type of join ('left', 'right', 'inner', 'outer')

- `lsuffix` → suffix to add if column names clash

- `rsuffix` → suffix for the other DataFrame


Types of Join
 
- inner:- Inner join returns only the rows that have matching keys in both DataFrames.


```python
import pandas as pd

# Left DataFrame
df1 = pd.DataFrame({'ID':[1,2,3], 'Name':['Alice','Bob','Charlie']})

# Right DataFrame
df2 = pd.DataFrame({'ID':[2,3,4], 'Age':[25,30,35]})

# Merge using inner join
result = pd.merge(df1, df2, on='ID', how='inner')
print(result)
```

What is left Join
- Left join returns all rows from the left DataFrame and only matching rows from the right DataFrame.

```python
import pandas as pd

# Left DataFrame
df1 = pd.DataFrame({'ID':[1,2,3], 'Name':['Alice','Bob','Charlie']})

# Right DataFrame
df2 = pd.DataFrame({'ID':[2,3,4], 'Age':[25,30,35]})

# Left join
result = pd.merge(df1, df2, on='ID', how='left')
print(result)
```

Left Join with Index (join method)

```python
df1 = pd.DataFrame({'Name':['Alice','Bob','Charlie']}, index=[1,2,3])
df2 = pd.DataFrame({'Age':[25,30,35]}, index=[2,3,4])

# Left join on index
result = df1.join(df2, how='left')
print(result)
```

What is Right Join
- Right join returns all rows from the right DataFrame and only matching rows from the left DataFrame.

```python
import pandas as pd

# Left DataFrame
df1 = pd.DataFrame({'ID':[1,2,3], 'Name':['Alice','Bob','Charlie']})

# Right DataFrame
df2 = pd.DataFrame({'ID':[2,3,4], 'Age':[25,30,35]})

# Right join
result = pd.merge(df1, df2, on='ID', how='right')
print(result)
```

What is Outer Join
- Outer join returns all rows from both DataFrames.

```python
import pandas as pd

# Left DataFrame
df1 = pd.DataFrame({'ID':[1,2,3], 'Name':['Alice','Bob','Charlie']})

# Right DataFrame
df2 = pd.DataFrame({'ID':[2,3,4], 'Age':[25,30,35]})

# Outer join
result = pd.merge(df1, df2, on='ID', how='outer')
print(result)
```
