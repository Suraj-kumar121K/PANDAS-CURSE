# 🟢 Handling Missing Data in Pandas
1. Missing data means kisi column ya row me value ka na hona.

2. Pandas me missing values ko mostly NaN (Not a Number) se represent kiya jata hai.

Missing values usually appear as:
- `NaN`  --> Numeric missing value
- `None` --> Python ka null value
- `NaT`  --> Date/Time missing value

```python
import pandas as pd
df = pd.read_excel(r"C:\EXCEL\converted_excel.xlsx")
df
```

# 1️⃣ Detect Missing Data
- Detect Missing Data means identifying the values in a dataset that are missing or null.

# 🔹 What .isnull() / isna()
- `isnull()` is used to detect missing values in a Pandas DataFrame.

- `isna()` same as `isnull()`

1. 🔹 Check missing values
- `df.isna()`
- `df.isnull()`   # same as isna()

```python
df.isnull()
```

# 🔹 Count missing values
`df.isna().sum()` is used to count missing values in each column of a Pandas DataFrame.
- `df.isna().sum()`


```python
df.isnull().sum()
```

# 2️⃣ Remove Missing Data
Remove Missing Data means deleting rows or columns that contain missing values (`NaN, None, NaT`) from a dataset.

### 📌 dropna() Method
dropna() is used to remove missing (null) values from a Pandas DataFrame or Series.

### 🔹 Syntax
df.dropna()

```python
df.dropna()
```

1️⃣ Remove columns with missing values
- `df.dropna(axis=1)`

```python
df.dropna(axis=0)
```

 2️⃣ Remove rows where all values are missing
- `df.dropna(how='all')`

```python
df.dropna(how="all")
```

3️⃣ Remove rows based on specific columns
- `df.dropna(subset=['Column1', 'Column2'])`

```python
df.dropna("SYMBOL", "TODAY-VOLUME", "AVG_VOLUME")
```

4️⃣ Apply changes to original DataFrame
- `df.dropna(inplace=True)`

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

# 3️⃣ Fill Missing Data
Fill Missing Data means replacing the missing or null values (`NaN, None, NaT`) in a dataset with meaningful values instead of removing them.

### 📌 `fillna()` Method
- fillna() is used to fill (replace) missing values in a Pandas DataFrame or Series.

### 🔹 Syntax
`df.fillna(value)`

```python
df.fillna({"SYMBOL":"PYTHON","SECURITY":"C","TODAY-VOLUME":"jAVA"})
```

#### 🔹 Common Examples
1️⃣ Fill with a constant value :- 
- `df.fillna(0)`

2️⃣ Fill a specific column :- 
- `df['Age'].fillna(25)`

3️⃣ Fill with mean value
- `df['Marks'].fillna(df['Marks'].mean())`

4️⃣ Forward fill (previous value)
- `df.fillna(method='ffill')`

5️⃣ Backward fill (next value)
- `df.fillna(method='bfill')`

6️⃣ Change original DataFrame :-  
- `df.fillna(0, inplace=True)`

# 5️⃣ Interpolate Missing Data
Interpolate Missing Data means estimating and filling missing values using existing data patterns, instead of using a fixed value like 0 or mean.

### 🔹 Syntax
`df.interpolate()`

```python
ar = pd.read_excel(r"C:\EXCEL\Converted_Excel_File.xlsx")
ar
```

### 🔹 Common Interpolation Methods

1️⃣ Linear interpolation (default)
- `df.interpolate()`
- Fills missing values in a straight-line manner.

2️⃣ Time-based interpolation
- `df.interpolate(method='time')`
- Used when index is DateTime.

3️⃣ Forward / Backward direction
- `df.interpolate(limit_direction='forward')`
