# Group By

# 1️⃣ What is GroupBy?
GroupBy in pandas is used to:
- Group rows based on one or more columns (keys)
- Perform aggregate operations on each group (like sum, mean, count, etc.)

# 2️⃣ Syntax
`df.groupby('column_name').agg_function()`
- `column_name` → column(s) to group by (key)
- `agg_function()` → aggregation like `sum()`, `mean()`, `count()`, `max()`, `min()`

```python
import pandas as pd

data = {
    'Department': ['HR', 'HR', 'IT', 'IT', 'Sales', 'Sales'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'],
    'Salary': [50000, 60000, 70000, 80000, 55000, 65000]
}

df = pd.DataFrame(data)
df
```

```python
var = df.groupby("Department")
var
```

```python
for x, y in var:
    print(x)
    print(y)
```

# 1️⃣ What is get_group?
- `groupby()`, pandas creates a GroupBy object.

- `get_group(key)` allows you to extract a single group from that object as a DataFrame.

```python
var.get_group("IT")
```

```python
var.min()
```

```python
var.max()
```
