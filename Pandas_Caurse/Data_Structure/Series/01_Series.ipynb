```python
import pandas as pd
import numpy as np
```

1. Series from lists

```python
data = ['india', 'Pakistan', 'USA', 'Nepal']
df = pd.Series(data, index=["A", "B", "C", "D"], name="Cuntry Name")
df
```

3. integers

```python
runs = [12, 123, 13, 45]
pd.Series(runs)
```

Custom Index

```python
data_01 = [67, 57, 89, 100]
subject = ['Math', 'English', 'Science', 'Hindi']
df = pd.Series(data_01, index = subject)
df
```

3. setting name

```python
df = pd.Series(data_01, index = subject, name="Suraj Marks")
df
```

Series from dict

```python
dict = {
    'Math':67,
    'English':57,
    'science':89,
    'Hindi':100
}
marks_series = pd.Series(dict, name="Suraj All Marks ")
marks_series
```
