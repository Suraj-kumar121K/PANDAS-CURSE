# 1️⃣ Filtering Data in pandas

# 📌 What is Filtering Data?
- Filtering means selecting specific rows or columns based on conditions.

1️⃣ Filter Rows using Condition
- DataFrame se sirf wahi rows select karna jo kisi condition ko satisfy karti ho.

🧠 Syntax

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'D'],
    'Marks': [35, 55, 70, 40]
})

print(df[df['Marks'] > 40])
```

# Multiple Conditions (AND / OR)

🔹 AND (`&`)

```python
df[(df['Marks'] > 40) & (df['Marks'] < 70)]
```

🔹 OR (`|`)

```python
df[(df['Marks'] < 40) | (df['Marks'] > 60)]
```

### 🔹 What is isin() in Pandas?
isin() ka use hota hai check karne ke liye

- 👉 koi value given list / set / array ke andar hai ya nahi.

🔹 Filter using isin()

```python
df[df['Name'].isin(['A', 'C'])]
```

Filter using between()

```python
df[df['Marks'].between(40, 60)]
```

# 🔥 Filter using String Conditions

🔹 Starts with

```python
df[df['Name'].str.startswith('A')]
```

🔹 Contains

```python
df[df['Name'].str.contains('B')]
```

🪶 Filter Missing Values
- DataFrame se un rows ko select karna jisme missing data (`NaN` / `None`) ho ya na ho.

🔹 isnull()

```python
df[df['Marks'].isnull()]
```

🔹 What is `notnull()`
- `notnull()` ka use hota hai check karne ke liye
- value missing (NaN / None) nahi hai.

```python
df[df['Marks'].notnull()]
```
