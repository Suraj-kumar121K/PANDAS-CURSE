# 🔹 describe() in Pandas
- describe() is used to get a statistical summary of a DataFrame or Series.

### 1️⃣ Basic Syntax
- `df.describe()`

### 2️⃣ What Does `describe()` Show?
- numeric columns, it returns: 

-  Statistic  -->        Meaning
-    count    --> Number of non-null values
-    mean     --> Average
-    std      --> Standard deviation
-    min      --> Minimum value
-    25%      --> First quartile
-    50%      --> Median
-    75%      --> Third quartile
-    max      --> Maximum value

```python
var.describe()
```

# 🔹 `sort_index()` in Pandas
`sort_index()` is used to sort a DataFrame or Series by its index.

### 🔹 axis in Pandas
`axis` tells Pandas which direction to work on
- axis=0 --> Rows (vertical)
- axis=1 --> Columns (horizontal)

### 1️⃣ Basic Syntax
`df.sort_index()`

### 3️⃣ Sort in Descending Order
`df.sort_index(ascending=False)`

### 5️⃣ Inplace Sorting
`df.sort_index(inplace=True)`

```python
var.sort_index(axis=0,ascending=False)
```
